```
from math import sqrt,pi
S = float(input('Введіть площу кола: '))
r = sqrt(S/pi)
print ('Радіус дорівнює %6.3f см' %r)
```

```
from math import pi

S = float(input("Введіть площу круга S: "))

R = (S / pi) ** 0.5
print(f"Радіус круга = {R:.3f} см")

S_sphere = 4 * pi * R**2
print(f"Площа сфери = {S_sphere:.3f} см²")

V_sphere = (4/3) * pi * R**3
print(f"Об'єм кулі = {V_sphere:.3f} см³")
```

```
print("Переведення температури:")
print("1 — Фаренгейт → Цельсій")
print("2 — Цельсій → Фаренгейт")

choice = input("Оберіть варіант (1 або 2): ")

if choice == "1":
    F = float(input("Введіть температуру у Фаренгейтах: "))
    C = (F - 32) * 0.5556
    print(f"{F}°F = {C:.2f}°C")

elif choice == "2":
    C = float(input("Введіть температуру у Цельсіях: "))
    F = C * 1.8 + 32
    print(f"{C}°C = {F:.2f}°F")

else:
    print("Помилка: потрібно обрати 1 або 2.")
```
